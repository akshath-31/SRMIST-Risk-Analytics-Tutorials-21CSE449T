```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

df = pd.read_csv("fraud_oracle.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# T9 - Fraud Distribution
# Count fraudulent and non-fraudulent claims using FraudFound_P
fraud_counts = df['FraudFound_P'].value_counts()
print("--- Fraud Distribution ---")
print(f"Non-Fraudulent Claims (0): {fraud_counts.get(0, 0)}")
print(f"Fraudulent Claims (1): {fraud_counts.get(1, 0)}")
```

```python
# T9 - Fraud Pattern Analysis
# Analyse fraud rates using PoliceReportFiled, AccidentArea, and PastNumberOfClaims
print("--- Fraud Pattern Analysis ---")

print("Fraud Rate by Police Report Filed:")
police_analysis = df.groupby('PoliceReportFiled')['FraudFound_P'].mean() * 100
print(police_analysis)
print()

print("Fraud Rate by Accident Area:")
area_analysis = df.groupby('AccidentArea')['FraudFound_P'].mean() * 100
print(area_analysis)
print()

print("Fraud Rate by Past Number Of Claims:")
claims_analysis = df.groupby('PastNumberOfClaims')['FraudFound_P'].mean() * 100
print(claims_analysis)
```

```python
# T9 - Fraud Detection

fraud_analysis_df = df.copy()

# Rule: No police report + Rural accident = Suspicious
def classify_claim(row):
    if row['PoliceReportFiled'] == 'No' and row['AccidentArea'] == 'Rural':
        return 'Suspicious'
    return 'Normal'

fraud_analysis_df['Fraud_Flag'] = fraud_analysis_df.apply(classify_claim, axis=1)

print("--- Fraud Classification Examples ---")
print(fraud_analysis_df[
    ['AccidentArea', 'PoliceReportFiled', 'PastNumberOfClaims', 'Fraud_Flag']
].head(10))
```

```python
# T9 - Fraud Handling
# Count the Normal and Suspicious claims from fraud_analysis_df
flag_counts = fraud_analysis_df['Fraud_Flag'].value_counts()
print("--- Fraud Handling ---")
print(f"Normal Claims: {flag_counts.get('Normal', 0)}")
print(f"Suspicious Claims: {flag_counts.get('Suspicious', 0)}")
print()

# Short statement explaining the verification policy
print("Policy Notice: All claims flagged as 'Suspicious' are held for detailed manual verification and investigative audit prior to approval, whereas 'Normal' claims proceed to routine automated processing.")
```
