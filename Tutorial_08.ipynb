```python
# T8 - RSS Parsing Using feedparser
import feedparser

# URL of a publicly accessible RSS feed related to insurance news (Insurance Journal)
feed_url = "https://www.insurancejournal.com/rss/"

# Read and parse the RSS feed
feed = feedparser.parse(feed_url)

# Create an empty list to store the extracted RSS data
rss_data = []

# Extract and print details of the first 5 articles
print(f"Feed Title: {feed.feed.title}\n")

for entry in feed.entries[:5]:
    title = entry.get('title', 'No Title')
    link = entry.get('link', 'No Link')
    published = entry.get('published', 'No Date')

    # Append the extracted details as a dictionary
    rss_data.append({"title": title, "link": link, "published": published})

    print(f"Title: {title}")
    print(f"Link: {link}")
    print(f"Published: {published}")
    print("-" * 40)
```

```python
# T8 - HTML Parsing Using BeautifulSoup
import requests
from bs4 import BeautifulSoup

# URL of a publicly accessible insurance-related webpage
url = "https://www.insurancejournal.com/"

# Send a GET request to the webpage
response = requests.get(url)

# Initialize variable to store parsed data later
html_data = {}

# Check if the request was successful
if response.status_code == 200:
    # Parse the returned HTML
    soup = BeautifulSoup(response.text, 'html.parser')

    # Extract a few simple pieces of text (e.g., the main page title and the first 5 article headlines)
    print("--- Webpage Title ---")
    page_title = soup.title.text.strip() if soup.title else "No Title Found"
    print(page_title)
    print("\n--- First 5 Headlines ---")

    # Finding h2 or h3 heading tags representing headlines
    raw_headlines = soup.find_all(['h2', 'h3'], limit=5)
    headlines_list = []
    for idx, headline in enumerate(raw_headlines, 1):
        headline_text = headline.text.strip()
        headlines_list.append(headline_text)
        print(f"{idx}. {headline_text}")

    # Create a dictionary called html_data
    html_data = {"page_title": page_title, "headlines": headlines_list}
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
```

```python
# T8 - Pickling
import pickle

# 1. Create a dictionary containing the extracted data
insurance_data = {
    "rss_data": rss_data,
    "html_data": html_data
}

# Define the filename
file_name = "insurance_data.pkl"

# 2. Save (pickle) the dictionary to a binary file
with open(file_name, "wb") as file:
    pickle.dump(insurance_data, file)

print(f"Data successfully pickled and saved to {file_name}\n")

# 3. Load (unpickle) the dictionary back from the file
with open(file_name, "rb") as file:
    loaded_data = pickle.load(file)

# 4. Print the loaded RSS and HTML data
print("--- Loaded RSS Data ---")
for idx, item in enumerate(loaded_data["rss_data"], 1):
    print(f"{idx}. {item['title']}")
    print(f"   Link: {item['link']}")
    print(f"   Published: {item['published']}")
print("\n--- Loaded HTML Data ---")
print(f"Page Title: {loaded_data['html_data'].get('page_title', 'N/A')}")
print("Headlines:")
for idx, headline in enumerate(loaded_data["html_data"].get("headlines", []), 1):
    print(f"  {idx}. {headline}")

# 5. Confirmation message
print("\nSuccess: The data was successfully saved and reloaded using Python's pickle library!")
```
