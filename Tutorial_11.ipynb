## Cell 1 — Import Libraries

```python
# Cell 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cell 2: Load Dataset

df = pd.read_csv('hospital_patient_treatment_dataset.csv')
df.head()
```

## Cell 3 — Dataset Overview

```python
# Cell 3: Dataset Overview
print("Dataset Shape:", df.shape)
print("\nColumn Names and Data Types:")
print(df.dtypes)
print("\nMissing Values per Column:")
print(df.isnull().sum())
print("\nNumber of Duplicate Records:", df.duplicated().sum())
```

## Cell 4 — Treatment Cost and Outcome Analysis

```python
# Cell 4: Treatment Cost and Outcome Analysis
analysis_table = df.groupby('Treatment Type').agg(
    Average_Cost=('Treatment Cost', 'mean'),
    Average_Recovery_Score=('Recovery Score', 'mean'),
    Average_Hospital_Stay=('Hospital Stay (Days)', 'mean')
).reset_index()

analysis_table
```

## Cell 5 — Treatment Cost Comparison

```python
# Cell 5: Treatment Cost Comparison
plt.figure(figsize=(10, 6))
sns.barplot(data=analysis_table, x='Treatment Type', y='Average_Cost', hue='Treatment Type', palette='viridis', legend=False)
plt.title('Average Treatment Cost by Treatment Type')
plt.xlabel('Treatment Type')
plt.ylabel('Average Cost (dataset units)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Cell 6 — Recovery Outcome Comparison

```python
# Cell 6: Recovery Outcome Comparison
plt.figure(figsize=(10, 6))
sns.barplot(data=analysis_table, x='Treatment Type', y='Average_Recovery_Score', hue='Treatment Type', palette='magma', legend=False)
plt.title('Average Recovery Score by Treatment Type')
plt.xlabel('Treatment Type')
plt.ylabel('Average Recovery Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Cell 7 — Cost vs Recovery Analysis

```python
# Cell 7: Cost vs Recovery Analysis
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Treatment Cost', y='Recovery Score', hue='Treatment Type', alpha=0.7)
plt.title('Relationship Between Treatment Cost and Recovery Score')
plt.xlabel('Treatment Cost (dataset units)')
plt.ylabel('Recovery Score')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Cell 8: Findings and Conclusion
print("Summary of Actual Findings (Descriptive Comparisons Only):")
print("1. Surgery has the highest average cost (~83,729.88 units) and highest average recovery score (~73.29).")
print("2. Therapy has the lowest average cost (~75,643.99 units).")
print("3. Observation has a higher average recovery score (~70.02) and lower average cost (~78,660.43 units) compared to Medication (~67.58 recovery, ~80,753.25 units).")
print("4. Surgery has the shortest average hospital stay (~15.21 days).")
print("\nCaution: These findings represent simple descriptive comparisons from the provided sample dataset and should not be used to infer verified clinical research or causal relationships.")
```
